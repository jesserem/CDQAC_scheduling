```python
import pandas as pd
import os
import numpy as np
```

```python
path_10_5_random = [
    "./checkpoints_exp/dataset_exp_10_5/CDQAC/sd1/random-99b991b0_seed_3",
    "./checkpoints_exp/dataset_exp_10_5/CDQAC/sd1/random-701eafec_seed_1",
    "./checkpoints_exp/dataset_exp_10_5/CDQAC/sd1/random-cb239ad6_seed_0",
    "./checkpoints_exp/dataset_exp_10_5/CDQAC/sd1/random-def9b02a_seed_2",
    ]

path_20_10_random = [
    "./checkpoints_exp/dataset_exp_20_10/CDQAC/sd1/random-1c46d5da_seed_1",
    "./checkpoints_exp/dataset_exp_20_10/CDQAC/sd1/random-2f4fb69b_seed_3",
    "./checkpoints_exp/dataset_exp_20_10/CDQAC/sd1/random-4ca44a74_seed_0",
    "./checkpoints_exp/dataset_exp_20_10/CDQAC/sd1/random-598fae9b_seed_2",
    ]

dict_res_30 = {
    "n_j": [10, 20],
    "n_m": [5, 10],
    "mean_makespan_greedy": [],
    "std_makespan_greedy": [],
    "mean_gap_greedy": [],
    "std_gap_greedy": [],
    "mean_runtime_greedy": [],
    "std_runtime_greedy": [],
    "mean_makespan_samp": [],
    "std_makespan_samp": [],
    "mean_gap_samp": [],
    "std_gap_samp": [],
    "mean_runtime_samp": [],
    "std_runtime_samp": [],
}
dict_res_40 = {
    "n_j": [10, 20],
    "n_m": [5, 10],
    "mean_makespan_greedy": [],
    "std_makespan_greedy": [],
    "mean_gap_greedy": [],
    "std_gap_greedy": [],
    "mean_runtime_greedy": [],
    "std_runtime_greedy": [],
    "mean_makespan_samp": [],
    "std_makespan_samp": [],
    "mean_gap_samp": [],
    "std_gap_samp": [],
    "mean_runtime_samp": [],
    "std_runtime_samp": [],
}

def get_mean_results(df_paths):
    makespan_greedy_30 = []
    gap_greedy_30 = []
    runtime_greedy_30 = []
    makespan_samp_30 = []
    gap_samp_30 = []
    runtime_samp_30 = []

    makespan_greedy_40 = []
    gap_greedy_40 = []
    runtime_greedy_40 = []
    makespan_samp_40 = []
    gap_samp_40 = []
    runtime_samp_40 = []

    for path in df_paths:
        df_30 = pd.read_csv(os.path.join(path, "SD1_30_10.csv"))
        df_40 = pd.read_csv(os.path.join(path, "SD1_40_10.csv"))
        mean_makespan_greedy_30 = df_30["makespan_det_offline"].mean()
        mean_gap_greedy_30 = df_30["gap_det_offline"].mean()
        mean_runtime_greedy_30 = df_30["runtime_det_offline"].mean()
        mean_makespan_samp_30 = df_30["makespan_stoch_offline"].mean()
        mean_gap_samp_30 = df_30["gap_stoch_offline"].mean()
        mean_runtime_samp_30 = df_30["runtime_stoch_offline"].mean()
        makespan_greedy_30.append(mean_makespan_greedy_30)
        gap_greedy_30.append(mean_gap_greedy_30)
        runtime_greedy_30.append(mean_runtime_greedy_30)
        makespan_samp_30.append(mean_makespan_samp_30)
        gap_samp_30.append(mean_gap_samp_30)
        runtime_samp_30.append(mean_runtime_samp_30)
        mean_makespan_greedy_40 = df_40["makespan_det_offline"].mean()
        mean_gap_greedy_40 = df_40["gap_det_offline"].mean()
        mean_runtime_greedy_40 = df_40["runtime_det_offline"].mean()
        mean_makespan_samp_40 = df_40["makespan_stoch_offline"].mean()
        mean_gap_samp_40 = df_40["gap_stoch_offline"].mean()
        mean_runtime_samp_40 = df_40["runtime_stoch_offline"].mean()
        makespan_greedy_40.append(mean_makespan_greedy_40)
        gap_greedy_40.append(mean_gap_greedy_40)
        runtime_greedy_40.append(mean_runtime_greedy_40)
        makespan_samp_40.append(mean_makespan_samp_40)
        gap_samp_40.append(mean_gap_samp_40)
        runtime_samp_40.append(mean_runtime_samp_40)
    m_makespan_greedy_30 = np.mean(makespan_greedy_30)
    s_makespan_greedy_30 = np.std(makespan_greedy_30)
    m_gap_greedy_30 = np.mean(gap_greedy_30)
    s_gap_greedy_30 = np.std(gap_greedy_30)
    m_runtime_greedy_30 = np.mean(runtime_greedy_30)
    s_runtime_greedy_30 = np.std(runtime_greedy_30)
    m_makespan_samp_30 = np.mean(makespan_samp_30)
    s_makespan_samp_30 = np.std(makespan_samp_30)
    m_gap_samp_30 = np.mean(gap_samp_30)
    s_gap_samp_30 = np.std(gap_samp_30)
    m_runtime_samp_30 = np.mean(runtime_samp_30)
    s_runtime_samp_30 = np.std(runtime_samp_30)

    m_makespan_greedy_40 = np.mean(makespan_greedy_40)
    s_makespan_greedy_40 = np.std(makespan_greedy_40)
    m_gap_greedy_40 = np.mean(gap_greedy_40)
    s_gap_greedy_40 = np.std(gap_greedy_40)
    m_runtime_greedy_40 = np.mean(runtime_greedy_40)
    s_runtime_greedy_40 = np.std(runtime_greedy_40)
    m_makespan_samp_40 = np.mean(makespan_samp_40)
    s_makespan_samp_40 = np.std(makespan_samp_40)
    m_gap_samp_40 = np.mean(gap_samp_40)
    s_gap_samp_40 = np.std(gap_samp_40)
    m_runtime_samp_40 = np.mean(runtime_samp_40)
    s_runtime_samp_40 = np.std(runtime_samp_40)


    dict_res_30["mean_makespan_greedy"].append(m_makespan_greedy_30)
    dict_res_30["std_makespan_greedy"].append(s_makespan_greedy_30)
    dict_res_30["mean_gap_greedy"].append(m_gap_greedy_30)
    dict_res_30["std_gap_greedy"].append(s_gap_greedy_30)
    dict_res_30["mean_runtime_greedy"].append(m_runtime_greedy_30)
    dict_res_30["std_runtime_greedy"].append(s_runtime_greedy_30)
    dict_res_30["mean_makespan_samp"].append(m_makespan_samp_30)
    dict_res_30["std_makespan_samp"].append(s_makespan_samp_30)
    dict_res_30["mean_gap_samp"].append(m_gap_samp_30)
    dict_res_30["std_gap_samp"].append(s_gap_samp_30)
    dict_res_30["mean_runtime_samp"].append(m_runtime_samp_30)
    dict_res_30["std_runtime_samp"].append(s_runtime_samp_30)

    dict_res_40["mean_makespan_greedy"].append(m_makespan_greedy_40)
    dict_res_40["std_makespan_greedy"].append(s_makespan_greedy_40)
    dict_res_40["mean_gap_greedy"].append(m_gap_greedy_40)
    dict_res_40["std_gap_greedy"].append(s_gap_greedy_40)
    dict_res_40["mean_runtime_greedy"].append(m_runtime_greedy_40)
    dict_res_40["std_runtime_greedy"].append(s_runtime_greedy_40)
    dict_res_40["mean_makespan_samp"].append(m_makespan_samp_40)
    dict_res_40["std_makespan_samp"].append(s_makespan_samp_40)
    dict_res_40["mean_gap_samp"].append(m_gap_samp_40)
    dict_res_40["std_gap_samp"].append(s_gap_samp_40)
    dict_res_40["mean_runtime_samp"].append(m_runtime_samp_40)
    dict_res_40["std_runtime_samp"].append(s_runtime_samp_40)



get_mean_results(path_10_5_random)
get_mean_results(path_20_10_random)
df_30 = pd.DataFrame(dict_res_30)
df_40 = pd.DataFrame(dict_res_40)
print(df_40.round(2))
# print(df_40)
```
